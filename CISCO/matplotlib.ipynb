## Matplotlib's Pyplot
Matplotlib.pyplot is Python's go-to library for data visualization. Ready to turn data into charts and graphs? Let's dive in and start plotting!



```python
# importing matplotlib as pyplot
import pandas as pd
import matplotlib.pyplot as plt
```

Most of the plotting methods take x and y values as their first two arguments. To illustrate, let's plot a simple up/down zig-zag pattern:`

```python
# Define x and y values.
# Plot x and y.
# Show the plot.

x = [1,2,3,4,5,6]
y = [1,8,1,8,1,8]
plt.plot(x,y)
plt.show()
```

```python
# Simple Increasing Line
x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.plot(x, y)
plt.show()
```

```python
# Multiple Lines in One Plot
x = [1,2,3,4,5]

y1 = [1,2,3,4,5]
y2 = [2,3,4,5,6]

plt.plot(x, y1)
plt.plot(x, y2)

plt.show()
```

```python
x = [0, 50, 100]
y = [10, 1, 4]
plt.plot(x, y, color='red')
```

The four basic plots

```python
# line graph
a = [1, 2, 3, 4]
b = [1, 4, 3, 4]
plt.plot(a, b)
```

```python
# bar plot
a = ['A','B','C']
b = [4, 1, 2]
plt.bar(a, b)
```

```python
# horizontal plot
a = ['A','B','C']
b = [4, 1, 2]
plt.barh(a, b)
```

```python
# scatter plot
a = [1, 2, 3, 4]
b = [1, 4, 3, 4]
plt.scatter(a, b)
```

Plotting DataFrames
Matplotlib integrates seamlessly with Pandas, making it straightforward to turn dataframes into visualizations. Let's see this in action using 'earth-layers.csv' from our previous project. First, load the data into a dataframe.

```python
df
```

```python
# Now let's plot the thickness of each layer as a bar chart.
# Use df['layer'] for the x values and df['thickness'] for y.
x = df['layer']
y = df['thickness']
plt.bar(x,y)
plt.show()
```

```python
# Labels and titles
# Matplotlib also gives us methods to add labels and titles. Try adding a title and ylabel to our bar chart.

plt.title('Layers of Earth')
plt.ylabel('Thicknes (km)')
plt.barh( df['layer'], df['thickness'])

# Notice how variables are not necessary. We were able to pass dataframe columns directly into the bar() method.
```
