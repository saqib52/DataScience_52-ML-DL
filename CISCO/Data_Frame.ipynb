Creating a dtaframe for:
Assign a color column.
Assign a radius column.
Show the result.

```python
df = pd.DataFrame()
df['color'] = ['blue', 'red', 'green', 'orange', 'yellow']
df['radius'] = [2,4,5,6,7]
df
```

```python
df['diameter'] = df['radius']*2
df
```

```python
df1 = pd.DataFrame()
df1['Fruits'] = ['Apple', 'Orange', 'bananna']
df1['Price'] =  [50,60,70]
df1['Quantity in Kg'] = [1,2,3]
df1
```

```python
df2 = pd.DataFrame()
df2['name'] = ['Ali', 'Ahmed', 'Sara', 'Zain', 'Ayesha']
df2['age'] = [20, 21, 19, 22, 20]
df2['marks'] = [85, 78, 92, 74, 88]
df2
```

```python
# Another way (Best Practice)
data = {'color': ['blue', 'red', 'green', 'orange', 'yellow'],'radius': [2, 4, 5, 6, 7]}
df= pd.DataFrame(data)
df
```
