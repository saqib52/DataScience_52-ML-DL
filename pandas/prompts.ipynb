# ✅ EDA Anomaly Checklist (Professional / Research-Ready)

Use this checklist during **every Exploratory Data Analysis (EDA)** before modeling, reporting, or decision-making.

---

## 1️⃣ Data Quality Anomalies (ALWAYS CHECK FIRST)

- [ ] Missing values (NaN, None, blanks)
- [ ] Missing values encoded as special numbers (0, -1, 999, -999)
- [ ] Duplicate rows or duplicate IDs
- [ ] Inconsistent data types (numbers stored as strings)
- [ ] Mixed units (e.g., kg vs g, °C vs °F)
- [ ] Impossible values (negative age, negative counts)
- [ ] Truncated or clipped values
- [ ] Rounding or precision loss
- [ ] Inconsistent categorical labels (e.g., `Male`, `male`, `M`)

**Action:** Remove / correct / flag  
**Risk if ignored:** Misleading summaries, broken models

---

## 2️⃣ Statistical Anomalies

- [ ] Extreme outliers
- [ ] Heavy skewness
- [ ] Non-normal distributions
- [ ] Zero-inflated variables
- [ ] Constant or near-constant columns
- [ ] Unrealistic variance
- [ ] Sudden distribution shifts

**Action:** Investigate meaning → transform or flag  
**Risk if ignored:** Distorted averages, unstable ML models

---

## 3️⃣ Structural / Schema Anomalies

- [ ] Unexpected number of columns
- [ ] Column meaning unclear or undocumented
- [ ] Same feature duplicated under different names
- [ ] Identifier fields mistakenly used as features
- [ ] One-hot encoded features mixed with raw categories
- [ ] Wide vs long format mismatch

**Action:** Redesign schema / restructure  
**Risk if ignored:** Leakage, incorrect feature importance

---

## 4️⃣ Logical & Consistency Anomalies

- [ ] Date inconsistencies (end date before start date)
- [ ] Aggregates not matching components
- [ ] Ratios > 100% or < 0%
- [ ] Mutually exclusive categories appearing together
- [ ] Violations of domain rules

**Action:** Validate against domain logic  
**Risk if ignored:** Invalid business or scientific conclusions

---

## 5️⃣ Temporal / Time-Series Anomalies

- [ ] Missing timestamps
- [ ] Irregular time intervals
- [ ] Sudden spikes or drops
- [ ] Seasonality not aligned with expectations
- [ ] Time leakage (future data influencing past)
- [ ] Trend breaks caused by external events

**Action:** Segment time, flag regime changes  
**Risk if ignored:** Over-optimistic forecasting

---

## 6️⃣ Relationship & Dependency Anomalies

- [ ] Unexpected correlations
- [ ] Multicollinearity
- [ ] Simpson’s paradox
- [ ] Nonlinear relationships hidden in aggregates
- [ ] Feature interactions ignored

**Action:** Stratify, transform, or model interactions  
**Risk if ignored:** False causality, misleading insights

---

## 7️⃣ Sampling & Bias Anomalies

- [ ] Class imbalance
- [ ] Underrepresented groups
- [ ] Biased data collection
- [ ] Convenience sampling
- [ ] Survivorship bias
- [ ] Label bias

**Action:** Reweight, resample, or disclose  
**Risk if ignored:** Ethical, legal, and performance failures

---

## 8️⃣ Business / Domain-Specific Anomalies

### Business / Marketing
- [ ] Revenue spikes due to one-time events
- [ ] Promotions masquerading as organic growth
- [ ] Customer churn caused by policy changes
- [ ] Fraud-like transaction patterns

### Biological / Scientific
- [ ] Batch effects
- [ ] Technical artifacts
- [ ] Outlier samples due to experimental failure
- [ ] Measurement saturation
- [ ] Platform-specific biases

**Action:** Flag or adjust using domain knowledge  
**Risk if ignored:** False insights, rejected papers, wrong decisions

---

## 9️⃣ ML-Specific Red Flags

- [ ] Target leakage
- [ ] Data leakage across train/test
- [ ] Features created using future information
- [ ] High accuracy with unrealistic logic
- [ ] Model performance collapsing on validation

**Action:** Rebuild pipeline  
**Risk if ignored:** Models that fail in production

---

## 🔟 EDA Final Sanity Questions (Ask Yourself)

- [ ] Do I trust this dataset?
- [ ] What assumptions am I making?
- [ ] What could go wrong?
- [ ] What should I explain to a client or reviewer?
- [ ] What would I double-check if money or funding depended on this?

---

## 🧠 Guiding Principle

> **EDA is not about plots.  
> It is about preventing wrong decisions.**

Always document anomalies you find — even if you decide not to fix them.
