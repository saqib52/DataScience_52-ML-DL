### How to do Skimming of the data

# What EDA really is (not the buzzword version)

Exploratory Data Analysis (EDA) is the process of understanding reality hidden inside raw data before you touch models, dashboards, or decisions.
Clients don’t pay for algorithms.They pay for clarity, confidence, and correct decisions.
EDA is how you get there.

# The Purpose of EDA (Client-Centric View)

1-To Understand What the Data Actually Represents

Clients often think they know their data.
They usually don’t.

Example:
Column: temperature
Client says: “This is daily temperature”\
EDA reveals:\
Some values are in °C, some in °F
Some are sensor errors (-999, 0)
Some days are missing

👉 Without EDA, your model is garbage in → garbage out.
<br>
EDA answers:\
1.What does each column really mean?\
2.What unit, scale, and range does it live in?\
3.Is this column even usable?

2- To Detect Data Quality Problems\
In real-world projects:
70–80% of effort is data cleaning, Clients rarely mention problems upfront\ 
EDA helps you find:
Missing values,
Duplicates,
Outliers,
Inconsistent categories,
Impossible values

Example (ecological data):\
Species count = -5,
Rainfall = 5000 mm/day,
Population = 0 for an inhabited region,\
These aren’t “small issues.” They change conclusions.

👉 Clients love EDA because it protects them from wrong decisions.

```python
# converting k into M for size column
df_GPS = pd.read_csv('./data/googleplaystore.csv', dtype={'Size': str})
```

```python
# Ensure Size is string
# Series object ke paas .strip() method nahi hota. .strip() sirf Python string object ke liye hota hai.
df_GPS['Size'] = df_GPS['Size'].astype(str).str.strip()
```

```python
# Replace 'Varies with device'
df_GPS['Size'] = df_GPS['Size'].replace('Varies with device', np.nan)
```

```python
# Extract numbers Basically, text me se numeric value extract karna.
size_num = df_GPS['Size'].str.extract(r'([\d.]+)')[0]

# .str.extract() → Pandas string method hai jo regular expression (regex) use karta hai aur string se pattern extract karta hai.
# r'([\d.]+)' → ye regex pattern hai.
# \d → kisi bhi digit ko match karta hai (0–9)
# . → dot . ko match karta hai (decimal numbers ke liye)
# [\d.]+ → ek ya zyada digits ya dots ko match karega
# () → parentheses ka matlab hai capture group, matlab ye wahi part return karega jo parentheses ke andar match hua

# ek line say dono tasks
# size_num = df_GPS['Size'].str.extract(r'([\d.]+)')[0].astype(float)
```

```python
# Detect units
is_mb = df_GPS['Size'].str.contains('M', na=False)
is_kb = df_GPS['Size'].str.contains('k', na=False)
```

```python
# Convert to MB
df_GPS['Size'] = np.where(is_mb,size_num.astype(float),np.where(is_kb,size_num.astype(float) / 1024,np.nan))
```
