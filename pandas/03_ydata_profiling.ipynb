##***Automation of EDA by Pandas-Profiling***\
Ydata-profiling (formerly known as pandas-profiling) is a powerful open-source Python library designed to automate Exploratory Data Analysis (EDA). It generates comprehensive, interactive HTML reports from Pandas or Spark DataFrames with a single line of code. 

#*Key Features*

#*Key features*
Type inference: automatic detection of columns' data types (Categorical, Numerical, Date, etc.)

Warnings: A summary of the problems/challenges in the data that you might need to work on (missing data, inaccuracies, skewness, etc.)

Univariate analysis: including descriptive statistics (mean, median, mode, etc) and informative visualizations such as distribution histograms

Multivariate analysis: including correlations, a detailed analysis of missing data, duplicate rows, and visual support for variables pairwise interaction

Time-Series: including different statistical information relative to time dependent data such as auto-correlation and seasonality, along ACF and PACF plots.

Text analysis: most common categories (uppercase, lowercase, separator), scripts (Latin, Cyrillic) and blocks (ASCII, Cyrilic)

File and Image analysis: file sizes, creation dates, dimensions, indication of truncated images and existence of EXIF metadata

Compare datasets: one-line solution to enable a fast and complete report on the comparison of datasets

Flexible output formats: all analysis can be exported to an HTML report that can be easily shared with different parties, as JSON for an easy integration in automated systems and as a widget in a Jupyter Notebook.

The report contains three additional sections:

Overview: mostly global details about the dataset (number of records, number of variables, overall missigness and duplicates, memory footprint)

Alerts: a comprehensive and automatic list of potential data quality issues (high correlation, skewness, uniformity, zeros, missing values, constant values, between others)

Reproduction: technical details about the analysis (time, version and configuration)

# ydata_profiling – Methods & Analysis Capabilities

This document summarizes the **methods, configuration options, and analysis capabilities** of the `ydata_profiling` library. It’s designed as a quick reference you can keep in your project or GitHub repo.

---

## 1. Core Object

```python
from ydata_profiling import ProfileReport
profile = ProfileReport(df)
```

Everything in `ydata_profiling` revolves around the **`ProfileReport`** object.

---

## 2. Main Public Methods

### 2.1 Generate a Report

```python
ProfileReport(df, title="EDA Report", explorative=True)
```

Creates a complete Exploratory Data Analysis (EDA) report.

---

### 2.2 Export Report to File

```python
profile.to_file("report.html")
```

Exports the report as an **HTML file** (PDF not supported directly).

---

### 2.3 Display in Jupyter Notebook

```python
profile.to_notebook_iframe()
```

Renders the report inside a Jupyter Notebook using an iframe.

---

### 2.4 Widget-Based View (Jupyter)

```python
profile.to_widgets()
```

Displays interactive widgets for notebook-based exploration.

---

### 2.5 Dataset Comparison (Data Drift)

```python
profile1.compare(profile2).to_file("comparison.html")
```

Used for:

* Before vs after cleaning
* Train vs test comparison
* Data drift detection

---

### 2.6 Programmatic Access to Results

```python
profile.get_description()
```

Returns a dictionary containing:

* Variable summaries
* Correlations
* Missing value stats
* Alerts

Useful for automation and ML pipelines.

---

## 3. Automatic Column Analysis

### 3.1 Numeric Variables

* Mean, median, standard deviation
* Min / max / range
* Variance
* Skewness & kurtosis
* Quantiles
* Zero & negative value counts
* Histograms and KDE plots

---

### 3.2 Categorical Variables

* Frequency tables
* Mode
* Cardinality
* Missing values
* Bar plots

---

### 3.3 Date & Time Variables

* Minimum & maximum date
* Date gaps
* Time distribution
* Seasonality detection

---

### 3.4 Text Variables

* Text length distribution
* Character frequency
* Word frequency
* Duplicate text detection
* Regex-based pattern detection

---

### 3.5 Image / File Path Variables

* File existence check
* Image resolution & size
* Duplicate image detection (hash-based)

---

## 4. Missing Value Analysis

Automatically includes:

* Missing count per column
* Missing percentage
* Missing value heatmap
* Missing value matrix

Configuration example:

```python
ProfileReport(df, missing_diagrams={"bar": True, "matrix": True})
```

---

## 5. Correlation Analysis

Supported correlation metrics:

| Method     | Use Case                     |
| ---------- | ---------------------------- |
| Pearson    | Linear numeric relationships |
| Spearman   | Monotonic relationships      |
| Kendall    | Rank-based correlation       |
| Cramér’s V | Categorical variables        |
| Phik       | Mixed data types             |

Enable correlations:

```python
ProfileReport(df, correlations={
    "pearson": {"calculate": True},
    "spearman": {"calculate": True},
    "kendall": {"calculate": True},
    "cramers": {"calculate": True},
    "phi_k": {"calculate": True}
})
```

---

## 6. Outlier Detection

Automatic detection using:

* IQR method
* Z-score

Displayed per column with summary alerts.

---

## 7. Duplicate & Uniqueness Analysis

* Duplicate rows
* Duplicate values per column
* Constant columns
* High cardinality warnings

---

## 8. Data Quality Alerts

Automatic alerts for:

* High missing percentage
* Highly correlated features
* Constant or near-constant columns
* Infinite values
* Highly skewed distributions

Minimal mode:

```python
ProfileReport(df, minimal=True)
```

---

## 9. Performance & Sampling Controls

For large datasets:

```python
ProfileReport(df, sample=10000, minimal=True)
```

---

## 10. Variable Type Overrides

Manually control variable interpretation:

```python
ProfileReport(
    df,
    variables={
        "age": {"type": "numeric"},
        "gender": {"type": "categorical"},
        "date": {"type": "datetime"}
    }
)
```

---

## 11. Advanced Configuration Options

### Explorative Mode

```python
ProfileReport(df, explorative=True)
```

Adds more plots, interactions, and deeper analysis.

---

### Interaction Analysis

```python
ProfileReport(df, interactions={"continuous": True})
```

---

### Sensitive Data Handling

```python
ProfileReport(df, sensitive=True)
```

---

## 12. When to Use ydata_profiling

Best suited for:

* Initial Exploratory Data Analysis (EDA)
* Data validation
* Quality checks
* Dataset comparison

Not intended for:

* Data cleaning
* Feature engineering
* Modeling

---

## 13. Typical Workflow

1. Load data with pandas
2. Generate profile report
3. Review alerts and distributions
4. Clean data using pandas
5. Compare before vs after reports

---

## 14. Summary

`ydata_profiling` provides:

* Descriptive statistics
* Distribution analysis
* Missing data analysis
* Correlation analysis
* Outlier detection
* Duplicate detection
* Data drift comparison
* Data quality alerts

It is an **EDA accelerator**, not a replacement for pandas or seaborn.
